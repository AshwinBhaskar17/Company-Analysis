```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_excel("/content/ABC Company.xlsx")
df
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
df.dtypes
```

```python
df["College"]=df["College"].fillna(df["College"].mode()[0])
df["College"]
```

```python
df["Salary"]=df['Salary'].fillna(df["Salary"].median())
df["Salary"]
```

```python
plt.figure(figsize=(10,5))
sns.boxplot(df)
```

```python
df.skew(numeric_only=True)
```

```python
Q1,Q3=np.percentile(df["Salary"],[25,75])
IQR=Q3-Q1
IQR
```

```python
lower_limit=Q1-1.5*IQR
```

```python
upper_limit=Q3+1.5*IQR
```

```python
df["Salary"]=df["Salary"].clip(lower_limit,upper_limit)
df["Salary"]
```

```python
Q1,Q3=np.percentile(df["Age"],[25,75])
IQR=Q3-Q1
IQR
```

```python
lower_limit=Q1-1.5*IQR
```

```python
upper_limit=Q3+1.5*IQR
```

```python
df["Age"]=df["Age"].clip(lower_limit,upper_limit)
df["Age"]
```

```python
plt.figure(figsize=(10,5))
sns.boxplot(df)
```

```python
df.hist()
```

```python
df.describe()
```

```python
df["Salary"]=np.log(df["Salary"]+1)
df["Salary"]
```

```python
np.random.seed(30)
df["Height"] = np.random.randint(150, 181, size=len(df))
df["Height"]
```

```python
df
```

```python
team_count = df["Team"].value_counts()
team_percentage = (team_count / len(df)*100).round(2)
team_count
```

```python
team_percentage
```

```python
team_count.plot(kind="bar", figsize=(12,5))
plt.title("Employees by Team")
plt.xlabel("Team")
plt.ylabel("Number of Employees")
plt.xticks(rotation=90)
plt.show()
```

'New Orleans Pelicans' this is the team has most employees
"Orlando Magic" "Minnesota Timberwolves" has the lowest


```python
position_count = df["Position"].value_counts()
position_count
```

```python
position_count.plot(kind="bar")
plt.title("Employees by Position")
plt.xlabel("Position")
plt.ylabel("Number of Employees")
plt.show()
```

employee is segregated by positions
SG has the highest and C has the lowest positions

```python
bins = [18, 22, 26, 30, 34, 38, 42]
labels = ["18-21", "22-25", "26-29", "30-33", "34-37", "38-41"]

df["Age_Group"] = pd.cut(df["Age"],bins=bins,labels=labels)
Age_Group=df["Age_Group"].value_counts().sort_index()
Age_Group
```

```python
df["Age"].max()
```

```python
Age_Group.plot(kind="bar")
plt.title("Employees by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Number of Employees")
plt.show()
```

the predominant age group is between 22-25 and the lowest is 38-41

```python
team_salary = df.groupby("Team")["Salary"].sum().sort_values(ascending=False)
team_salary
```

```python
team_salary = df.groupby(["Team","Position"])["Salary"].sum().sort_values(ascending=False)
team_salary
```

```python
position_salary
```

```python
position_salary = df.groupby("Position")["Salary"].sum()
plt.bar(position_salary.index, position_salary.values)
plt.title("Salary Expenditure by Position")
plt.xlabel("Position")
plt.ylabel("Total Salary")
plt.show()
```

```python
correlation = df["Age"].corr(df["Salary"])
correlation
```

```python
plt.scatter(df["Age"], df["Salary"])
plt.title("Age vs Salary")
plt.xlabel("Age")
plt.ylabel("Salary")
plt.show()
```

most of the people work between 20-30 and highest salary is also earned between this age. When age is more the salary is getting decreased. Most of the employees earn the salary between 20-35
